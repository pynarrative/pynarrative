```python
%load_ext autoreload
%autoreload 2
%reload_ext autoreload


import pynarrative as pn

import pandas as pd
import altair as alt

print("Percorso libreria caricata:", pn.__file__)
```

```python
import numpy as np
import string

np.random.seed(42)

len = 10
valori = np.random.randint(0, 100, len)

dati_categorie = pd.DataFrame({
    "categorie": [i for i in string.ascii_uppercase[:len]],
    "valori": valori
})


dati_categorie
```

*Historical Template*

```python
from pynarrative import HistoricalTemplate as HT

storia = (
    pn.Story(
    #Costruzione dell'oggetto di classe Story
        data = dati_categorie,
        width = 500,
        height = 300,
        template = HT
    )

    #Chiamata dei metodi

    .mark_bar(
        cornerRadiusTopLeft = 7,
        cornerRadiusTopRight = 7,

    )

    .encode( #encoding dei dati
        x = alt.X("categorie:N", title = "Categoria", axis = alt.Axis(grid = True)),
        y = alt.Y("valori:Q", title = "Valore", axis = alt.Axis(grid = True))
    )
    

    .add_title( #titolo e sottotitolo
        title = "Grafico a barre con dati e categorie",
        subtitle = "Esempio di utilizzo di pynarrative",
        align = "center"
    )

    .add_context( #contesto, testo, spiegazione
        text = "Lorem ipsum dolor sit amet, consectetuer adipiscing elit, sed diam nonummy nibh euismod tincidunt ut laoreet dolore magna aliquam erat volutpat. Ut wisi enim ad minim veniam, quis nostrud exerci tation ullamcorper suscipit lobortis nisl ut aliquip ex ea commodo consequat.",
        position = "left",
    )

    .add_context( #contesto, testo, spiegazione
        text = "Lorem ipsum dolor sit amet, consectetuer adipiscing elit, sed diam nonummy nibh euismod tincidunt ut laoreet dolore magna aliquam erat volutpat. Ut wisi enim ad minim veniam, quis nostrud exerci tation ullamcorper suscipit lobortis nisl ut aliquip ex ea commodo consequat.",
        position = "right",
    )

    .add_next_steps( #prossima parte della storia
        steps = ["NS1", "NS2"],
        title = "Next Steps", #
        position = "bottom" #
    )

    .add_source( #fonte384db5
        text = "Fonte: dati di esempio",
        position = "bottom",
        align = "right"
    )


    .add_labels_chart(
        font_size = 20,
        font_weight = "bold",
        dy = 20
    )

    .add_line(
        orientation = "horizontal",
        value = 50,
    )

    .add_highlight("B")
    


    .render()
)

storia
```

```python
np.random.seed(43)

len = 10

valori1 = np.random.randint(50, 100, len)
valori2 = np.random.randint(40, 150, len)
valori3 = np.random.randint(0, 60, len)
# valori4 = np.random.randint(0, 100, len)
# valori5 = np.random.randint(0, 100, len)

dati_categorie_multiserie = pd.DataFrame({
    "categorie": [i for i in string.ascii_uppercase[:len]],
    "valori1": valori1,
    "valori2": valori2,
    "valori3": valori3,
    # "valori4": valori4,
    # "valori5": valori5,
})


dati_categorie_multiserie
```

```python
dati_categorie_multiserie_long = dati_categorie_multiserie.melt(
    id_vars = "categorie",  # Colonna che rimane fissa
    value_vars = ["valori1", "valori2", "valori3"],#, "valori4", "valori5"],  # Colonne da trasformare
    var_name = "serie",  # Nome della nuova colonna categorica
    value_name = "valori"  # Nome della colonna con i valori
)

dati_categorie_multiserie_long.head()
```

```python
from pynarrative import BeigeBlueTemplate as BBT

storia = (
    pn.Story(
    #Costruzione dell'oggetto di classe Story
        data = dati_categorie_multiserie_long,
        width = 700,
        height = 400,
        template = BBT
    )

    #Chiamata dei metodi

    .mark_bar(
        # cornerRadiusTopLeft = 7,
        # cornerRadiusTopRight = 7,
        # size = 15
    )

    .encode(
        x = alt.X(
            "categorie:N",
            title = "Categoria",
            axis = alt.Axis(grid = True),
            scale = alt.Scale(paddingInner = 0.3) #Spazio tra i gruppi
            ),
        y = alt.Y("valori:Q", title = "Valore", axis = alt.Axis(grid = True)),
        color = alt.Color("serie:N", title = "Legenda"),

        xOffset = alt.XOffset(
            "serie:N",
            scale = alt.Scale(paddingInner = 0.1) #Spazio tra le barre del singolo gruppo
        )
    )
    

    .add_title( #titolo e sottotitolo
        title = "Grafico a barre con dati e categorie",
        subtitle = "Esempio di utilizzo di pynarrative",
        align = "center"
    )

    .add_context( #contesto, testo, spiegazione
        text = "Lorem ipsum dolor sit amet, consectetuer adipiscing elit, sed diam nonummy nibh euismod tincidunt ut laoreet dolore magna aliquam erat volutpat. Ut wisi enim ad minim veniam, quis nostrud exerci tation ullamcorper suscipit lobortis nisl ut aliquip ex ea commodo consequat.",
        position = "left",
        font_weight = "bold"
    )

    .add_context( #contesto, testo, spiegazione
        text = "Lorem ipsum dolor sit amet, consectetuer adipiscing elit, sed diam nonummy nibh euismod tincidunt ut laoreet dolore magna aliquam erat volutpat. Ut wisi enim ad minim veniam, quis nostrud exerci tation ullamcorper suscipit lobortis nisl ut aliquip ex ea commodo consequat.",
        position = "right",
    )

    .add_next_steps( #prossima parte della storia
        steps = ["NS1", "NS2"],
        title = "Next Steps", #
        position = "bottom" #
    )

    .add_source(
        text = "Fonte: dati di esempio",
        position = "bottom",
        align = "right"
    )


    .add_line(
        orientation = "horizontal",
        math = "mean",
        value = dati_categorie_multiserie["valori1"].tolist(),
        color = "darkblue",
        label_text = "Media serie 1 (71,2)",
        label_font_weight = "bold",
        label_font_size = 16,
    )

    .add_line(
        orientation = "horizontal",
        math = "mean",
        value = dati_categorie_multiserie["valori2"].tolist(),
        color = "darkorange",
        label_text = "Media serie 2 (104)",
        label_font_weight = "bold",
        label_font_size = 16
    )

    .add_line(
        orientation = "horizontal",
        math = "mean",
        value = dati_categorie_multiserie["valori3"].tolist(),
        color = "darkgreen",
        label_text = "Media serie 3 (34,4)",
        label_font_weight = "bold",
        label_font_size = 16
    )

    .add_highlight("C")


    .render()
)

storia
```

```python
np.random.seed(47)
valori1 = np.random.randint(50, 80, 20)
valori2 = np.random.randint(50, 100, 20)
valori3 = np.random.randint(50, 65, 20)

dati = pd.DataFrame({
    "valori1": valori1,
    "valori2": valori2,
    "valori3": valori3,
})

dati["indice"] = dati.index

dati = dati[["indice", "valori1", "valori2", "valori3"]]



dati.tail()
```

```python
dati_long = dati.melt(id_vars='indice', 
                       var_name='serie', 
                       value_name='valore')

dati_long.head()
```

```python
storia = (
    pn.Story(
    #Costruzione dell'oggetto di classe Story
        # data = dati_long,
        data = dati,
        width = 700,
        height = 450,
        template = BBT
    )

    #Chiamata dei metodi

    # .mark_point()
    .mark_line(
        point = True,
        interpolate = "cardinal",
        # interpolate = "basis",
        # interpolate = "step",
        # interpolate = "step-before",
        # interpolate = "step-after",
        strokeDash=[10, 4]
    )

    .encode(
        x = alt.X("indice:Q", title = "Categoria", axis = alt.Axis(grid = True, labelAngle = 0 )),
        # y = alt.Y("valore:Q", title = "Valore", axis = alt.Axis(grid = True)),
        y = alt.Y("valori1:Q", title = "Valore", axis = alt.Axis(grid = True), scale = alt.Scale(domain=[0, 100])),
        # color = alt.Color("serie:N", title = "Serie"),
    )
    

    .add_title( #titolo e sottotitolo
        title = "Grafico a barre con dati e categorie",
        subtitle = "Esempio di utilizzo di pynarrative",
        align = "center"
    )

    .add_context( #contesto, testo, spiegazione
        text = "Lorem ipsum dolor sit amet, consectetuer adipiscing elit, sed diam nonummy nibh euismod tincidunt ut laoreet dolore magna aliquam erat volutpat. Ut wisi enim ad minim veniam, quis nostrud exerci tation ullamcorper suscipit lobortis nisl ut aliquip ex ea commodo consequat.",
        position = "left",
    )

    .add_next_steps( #prossima parte della storia
        steps = ["NS1", "NS2"],
        title = "Next Steps", #
        position = "bottom" #
    )

    .add_source( #fonte
        text = "Fonte: dati di esempio",
        position = "bottom",
        align = "right"
    )


    .add_line(
        orientation = "horizontal",
        math = "mean",
        value = dati["valori1"]
    )

    .add_line(
        orientation = "horizontal",
        math = "min",
        value = dati["valori1"],
        color = "darkblue"
    )

    .add_line(
        orientation = "horizontal",
        math = "max",
        value = dati["valori1"],
        color = "darkgreen"
    )

    .render()
)

storia
```
