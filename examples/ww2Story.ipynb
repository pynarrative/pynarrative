**Creating dataframes**

Data source: *Source: Various sources, The National World War II Museum via statista.com*

```python
country = [
    "Soviet Union", "China", "Germany", "Poland", "Indonesia", "Japan",
    "India", "French Indochina", "Yugoslavia", "Romania", "Greece",
    "Philippines", "Hungary", "France", "Italy", "United Kingdom",
    "Korea", "United States", "Austria", "Lithuania", "Czechoslovakia",
    "Netherlands", "Burma (Myanmar)", "Latvia", "Ethiopia",
    "Malaya (Malaysia, Singapore)", "Finland", "Belgium", "Estonia",
    "Portuguese Timor (East Timor)", "Canada & Newfoundland", "Australia",
    "Albania", "Bulgaria", "Papua & New Guinea", "Spain", "New Zealand",
    "South Africa", "Norway", "Thailand", "Denmark", "Ireland", "Sweden",
    "Brazil", "Luxembourg", "Malta", "Egypt", "Iraq", "Nauru", "Mongolia",
    "Iceland", "Iran", "Turkey", "Cuba", "Mexico", "Switzerland"
]

death = [
    24000000, 19500000, 7700000, 5600000, 3500000, 2850000,
    2087000, 1250000, 1000000, 833000, 825000, 807000,
    580000, 567600, 457000, 450700, 425500, 418500,
    384700, 353000, 345000, 301000, 272000, 227000,
    100000, 100000, 96800, 86100, 51000, 50000,
    44090, 39654, 30200, 25000, 15000, 14500,
    11900, 11900, 9500, 7600, 6600, 5000,
    2100, 2000, 2000, 1581, 1100, 700,
    600, 300, 200, 200, 200, 100,
    100, 100
]

#Let's take the first 20 records
ww2_death = pd.DataFrame({
    "country" : country[:20],
    "death" : death[:20]
})

#Convert total death count to millions and round to 2 decimal places
ww2_death["death"] = round(ww2_death["death"] /1000000, 2)
```

Data story with **bar chart**

```python
story = (
    pn.Story(
    #Builing the Story class object
        data = ww2_death,
        width = 700,
        template = militaryTemplate #Using militaryTemplate
    )

    #Method invocation
    #Bar chart
    .mark_bar(
        cornerRadiusEnd = 7,
        size = 15
    )

    #Data encoding
    #To create a horizontal bar chart we plot the cathegorical data on Y and the numerical data on X
    .encode(
        x = alt.X(
            "country:N",
            title = "Country", 
            axis = alt.Axis(
                grid = True,
                labelAngle = -45
            ),
            sort = "-y" #Sorting the data in a descending order
        ),
        y = alt.Y(
            "death:Q",
            title = "Estimated number of deaths due to the Second World War (in millions)",
            axis = alt.Axis(
                grid = True
            ),
        ),
    )
    
    #Data source
    .add_source(
        text = "Source: Various sources, The National World War II Museum via statista.com",
        position = "top",
        align = "right"
    )
    
    #Title and subtitle
    .add_title(
        title = "Estimated number of fatalities due to the Second World War",
        subtitle = "In millions, by country or region, from 1939 to 1945, first 20 countries or regions",
        align = "center"
    )
    
    #Context area (on the right)
    .add_context(
        text = "World War II resulted in unprecedented human loss between 1939 and 1945, with fatalities heavily concentrated in Eastern Europe and East Asia. The Soviet Union and China suffered the vast majority of casualties, recording 24 million and 19.5 million deaths respectively. Germany (7.7M) and Poland (5.6M) follow as the most severely impacted nations in Central Europe, while Asian territories under occupation—such as the Dutch East Indies (3.5M) and Japan (2.85M)—also endured massive losses. Western nations like Italy (0.46M), the United Kingdom (0.45M), and the United States (0.42M) accounted for a comparatively smaller share of total mortality.",
        position = "left",
    )

    # Highlighting line (mean)
    .add_line(
        orientation = "horizontal",
        value = ww2_death["death"],
        math = "mean",
        label_text = "Average deaths in WW2: 3.68 millions",
        label_font_size = 13,
        color = "rgb(160, 175, 136)",
        label_font_weight = "bold"
    )

    .add_highlight(["Soviet Union", "United States"])

    .add_labels_chart(
        values = "death:Q",
        font_weight = "bold",
        font_size = 14,
        dy = -20
    )

    .render()
)

story
```
