**Creating dataframes**

Data source: *Source: Organization for Economic Co-operation and Development via statista.com*

```python
year = [
    1956, 1957, 1958, 1959, 1960, 1961, 1962, 1963, 1964, 1965, 1966, 1967,
    1968, 1969, 1970, 1971, 1972, 1973, 1974, 1975, 1976, 1977, 1978, 1979,
    1980, 1981, 1982, 1983, 1984, 1985, 1986, 1987, 1988, 1989, 1990, 1991,
    1992, 1993, 1994, 1995, 1996, 1997, 1998, 1999, 2000, 2001, 2002, 2003,
    2004, 2005, 2006, 2007, 2008, 2009, 2010, 2011, 2012, 2013, 2014, 2015,
    2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024, 2025
]

inflation_rate = [
    3.38, 1.32, 2.86, -0.51, 2.35, 2.05, 4.69, 7.45, 5.91, 4.52, 2.35, 3.73,
    1.28, 2.66, 4.97, 4.79, 5.75, 10.8, 19.16, 16.95, 16.61, 17.13, 12.09, 14.8,
    21.06, 17.97, 16.48, 14.65, 10.79, 9.21, 5.82, 4.75, 5.06, 6.26, 6.46, 6.25,
    5.27, 4.63, 4.05, 5.24, 4.01, 2.04, 1.96, 1.66, 2.54, 2.79, 2.47, 2.67,
    2.21, 1.99, 2.09, 1.83, 3.35, 0.77, 1.53, 2.78, 3.04, 1.22, 0.24, 0.04,
    -0.09, 1.23, 1.14, 0.61, -0.14, 1.87, 8.2, 5.62, 0.98, 1.53
]

inflation_italy = pd.DataFrame({
    "year" : year,
    "inflation_rate" : inflation_rate
})
```

Data story with **horizontal bar chart**

```python
story = (
    pn.Story(
    #Builing the Story class object
        data = inflation_italy,
        width = 700,
        height = 300,
        template = orangeTemplate #Using orangeTemplate
    )

    #Method invocation
    #Line chart
    .mark_line()

    #Data encoding
    #To create a horizontal bar chart we plot the cathegorical data on Y and the numerical data on X
    .encode(
        x = alt.X(
            # 'year' must be encoded as Quantitative (:Q) instead of Ordinal (:O) or Nominal (:N) because 
            # numerical features like scale limits (domain=[1956, 2025]), number formatting (format="d"), 
            # and precise axis label values (values = [y for y in range(1956, 2026, 4)]) require continuous numeric values.
            "year:Q",
            title = "",
            axis = alt.Axis(
                labelAngle = -30,
                format = "d",
                values = [y for y in range(1956, 2026, 4)]
            ),
            scale = alt.Scale(domain=[1956, 2025])
        ),
        y = alt.Y(
            "inflation_rate:Q",
            title = "Inflation rate (%)",
        ),
    )
    
    #Data source
    .add_source(
        text = "Source: Organisation for Economic Co-operation and Development via statista.com",
        position = "bottom",
        align = "left"
    )
    
    #Title and subtitle
    .add_title(
        title = "Italy’s Inflation Rollercoaster (1956–2025)",
        align = "left",
    )

    #Context area (on top)
    .add_context(
        text = "Between the late 1970s and early 1980s, Italy experienced historic double-digit inflation peaks, driven primarily by energy shocks and wage-price spirals. Following the entry into the European Monetary System and subsequent reforms, rates steadily declined, remaining largely subdued until the recent post-2021 global supply chain disruptions and energy crisis pushed figures upward once again.",
        position = "top",
        title = "",
        width_multiplier = 1.7
    )

    #Average line
    .add_line(
        value = inflation_italy["inflation_rate"],
        math = "mean",
        label_text = " ",
        color = "black",
        label_font_size = 15
    )

    .add_annotation(
        x = 1977,
        y = 22,
        text = "Severe inflationary pressures dominated the 1970s and 80s before structural reforms and the Adoption of the Euro stabilized rates near the 5.33% historical average.",
        dx = 80
    )

    .add_annotation(
        x = 2021,
        y = 8,
        text = "Covid 19 pandemic",
        dx = -75,
        dy = -60
    )


    #Next steps
    .add_next_steps(
        position = "right",
        mode = "vertical",
        title = "NEXT STEPS",
        steps = [
                "Implement targeted fiscal restraint to avoid overheating aggregate demand",
                "Diversify national energy supply to hedge against future fossil fuel price shocks",
                "Streamline market regulations to promote competition and lower consumer prices"
                ],
        text_align = "left",
        img_url = [
            "https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcQnhjS1BUzwWfkOr1FSkFLO7v4hcuo9fhl1YUTcdfaYGw&s=10",
            "https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcS7Cup3WdQ2jI-WJHzn5nELf7EZ2LjKHqHK6Ip3wyYXxA&s=10",
            "https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcQ20iy8VSOW_khlkEIHIc3lGtdaApDRhWcP8q_mJo-50Q&s=10"
            ],
        img_width = 50
    )


    .render()
)

story
```
